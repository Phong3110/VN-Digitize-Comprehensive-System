```python
from paddleocr import PaddleOCR
from vietocr.tool.predictor import Predictor
from vietocr.tool.config import Cfg
from PIL import Image
import numpy as np

# Load models
ocr = PaddleOCR(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
    text_detection_model_name="PP-OCRv5_server_det",
    ocr_version="PP-OCRv5",
    device="gpu",
    text_det_limit_side_len=960,
    text_det_limit_type="max",
    text_det_thresh=0.3,
    text_det_box_thresh=0.5,
    text_det_unclip_ratio=2.0,
)

config = Cfg.load_config_from_name('vgg_transformer')
config['device'] = 'cuda:0'
config['predictor']['beamsearch'] = False
predictor = Predictor(config)

img_np = np.array(Image.open("page1_preview.png").convert("RGB"))
img_h, img_w = img_np.shape[:2]

# Detection
results = ocr.predict(img_np)

# Tách từng box thành dòng đơn
def split_box_to_lines(img_np, poly, line_height_ratio=1.3):
    """
    Nếu box cao hơn 1 dòng → chia đều thành N dòng đơn.
    line_height_ratio: hệ số ước tính chiều cao 1 dòng
    """
    xs = [p[0] for p in poly]
    ys = [p[1] for p in poly]
    x1, y1 = int(min(xs)), int(min(ys))
    x2, y2 = int(max(xs)), int(max(ys))
    
    box_h = y2 - y1
    box_w = x2 - x1
    
    # Ước tính chiều cao 1 dòng dựa trên chiều rộng box
    # Chữ tiêu chuẩn A4: ratio height/width của 1 dòng ~ 0.035
    estimated_line_h = max(20, int(img_w * 0.030))
    
    n_lines = max(1, round(box_h / estimated_line_h))
    
    if n_lines == 1:
        return [(x1, y1, x2, y2)]
    
    # Chia đều thành n_lines
    line_h = box_h // n_lines
    lines = []
    for i in range(n_lines):
        ly1 = y1 + i * line_h
        ly2 = y1 + (i + 1) * line_h if i < n_lines - 1 else y2
        lines.append((x1, ly1, x2, ly2))
    return lines

# Crop và recognize từng dòng
all_lines = []
for res in results:
    for poly in res.get("rec_polys", []):
        polygon = poly.tolist()
        line_boxes = split_box_to_lines(img_np, polygon)
        for (x1, y1, x2, y2) in line_boxes:
            # Thêm padding nhỏ
            pad = 3
            x1c = max(0, x1-pad); y1c = max(0, y1-pad)
            x2c = min(img_w, x2+pad); y2c = min(img_h, y2+pad)
            crop = Image.fromarray(img_np[y1c:y2c, x1c:x2c])
            all_lines.append((crop, (x1, y1, x2, y2)))

print(f"Tổng dòng sau khi tách: {len(all_lines)}")

# Recognize batch
crops = [line[0] for line in all_lines]
texts = predictor.predict_batch(crops)

print("\n=== KẾT QUẢ ===")
for (crop, bbox), text in zip(all_lines, texts):
    print(f"  {text}")
```

```python
from paddleocr import PaddleOCR
from PIL import Image, ImageDraw
import numpy as np

ocr = PaddleOCR(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
    text_detection_model_name="PP-OCRv5_server_det",
    device="gpu",
    text_det_limit_side_len=960,
    text_det_limit_type="max",
    text_det_thresh=0.3,
    text_det_box_thresh=0.5,
    text_det_unclip_ratio=2.0,
)

img_np = np.array(Image.open("page1_preview.png").convert("RGB"))
results = ocr.predict(img_np)

# Lấy tất cả boxes
boxes = []
for res in results:
    for poly in res.get("rec_polys", []):
        p = poly.tolist()
        xs = [pt[0] for pt in p]; ys = [pt[1] for pt in p]
        boxes.append({
            "x1": int(min(xs)), "y1": int(min(ys)),
            "x2": int(max(xs)), "y2": int(max(ys)),
        })

# Tính IoU để tìm box trùng nhau
def iou(a, b):
    ix1 = max(a["x1"], b["x1"]); iy1 = max(a["y1"], b["y1"])
    ix2 = min(a["x2"], b["x2"]); iy2 = min(a["y2"], b["y2"])
    inter = max(0, ix2-ix1) * max(0, iy2-iy1)
    if inter == 0: return 0
    area_a = (a["x2"]-a["x1"]) * (a["y2"]-a["y1"])
    area_b = (b["x2"]-b["x1"]) * (b["y2"]-b["y1"])
    return inter / min(area_a, area_b)  # overlap ratio

# In các cặp box trùng nhau
print(f"Tổng boxes: {len(boxes)}")
print("\nCác box bị CHỒNG NHAU (overlap > 50%):")
for i in range(len(boxes)):
    for j in range(i+1, len(boxes)):
        score = iou(boxes[i], boxes[j])
        if score > 0.5:
            print(f"  Box {i} {boxes[i]} ←→ Box {j} {boxes[j]}  overlap={score:.0%}")

# Vẽ tất cả boxes để xem
img_draw = Image.fromarray(img_np.copy())
draw = ImageDraw.Draw(img_draw)
for b in boxes:
    draw.rectangle([b["x1"], b["y1"], b["x2"], b["y2"]], outline="red", width=2)
img_draw.save("page1_boxes_debug.png")
print("\nĐã lưu page1_boxes_debug.png")
```

```python
import fitz
import numpy as np
from PIL import Image, ImageDraw
from paddleocr import PaddleOCR

ocr = PaddleOCR(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
    text_detection_model_name="PP-OCRv5_server_det",
    device="gpu",
    text_det_limit_side_len=960,
    text_det_limit_type="max",
    text_det_thresh=0.3,
    text_det_box_thresh=0.5,
    text_det_unclip_ratio=2.0,
)

doc = fitz.open("pdf_image_only.pdf")
zoom = 200 / 72.0
mat = fitz.Matrix(zoom, zoom)

# Export trang 2, 3, 4 để kiểm tra
for page_num in [1, 2, 3]:  # 0-indexed
    pix = doc[page_num].get_pixmap(matrix=mat, alpha=False)
    img = np.frombuffer(pix.samples, dtype=np.uint8).reshape(pix.h, pix.w, 3)
    
    results = ocr.predict(img)
    img_draw = Image.fromarray(img.copy())
    draw = ImageDraw.Draw(img_draw)
    
    count = 0
    for res in results:
        for poly in res.get("rec_polys", []):
            p = poly.tolist()
            xs = [pt[0] for pt in p]; ys = [pt[1] for pt in p]
            x1,y1,x2,y2 = int(min(xs)),int(min(ys)),int(max(xs)),int(max(ys))
            draw.rectangle([x1,y1,x2,y2], outline="red", width=2)
            count += 1
    
    img_draw.save(f"page{page_num+1}_debug.png")
    print(f"Trang {page_num+1}: {count} boxes → lưu page{page_num+1}_debug.png")
```
