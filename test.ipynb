```python
from pdf2image import convert_from_path

# 1. Chuyển PDF thành danh sách các ảnh (Rasterize)
images = convert_from_path('test.pdf', dpi=300) # 300 DPI để giữ độ nét cho OCR

# 2. Lưu danh sách ảnh này thành 1 file PDF mới (dạng ảnh, không thể copy)
images[0].save('pdf_khong_copy_duoc.pdf', save_all=True, append_images=images[1:])

print("Đã tạo xong file PDF dạng ảnh để test OCR!")
```

```python
from pdf2image import convert_from_path
import os

# TRỎ ĐƯỜNG DẪN ĐẾN THƯ MỤC BIN CỦA POPPLER BẠN VỪA GIẢI NÉN
path_to_poppler = r"E:\poppler-windows\Library\bin" # Thay bằng đường dẫn thực tế của bạn

# Chuyển PDF thành ảnh
images = convert_from_path(
    'quetdc.pdf', 
    dpi=200,
    size=(800, None),
    poppler_path=path_to_poppler # Thêm dòng này để fix lỗi
)

# Lưu lại thành PDF không thể copy
images[0].save('pdf_image_only.pdf', save_all=True, append_images=images[1:])

print("Xong rồi Phong! Giờ file PDF đã là dạng ảnh 100%.")
```
